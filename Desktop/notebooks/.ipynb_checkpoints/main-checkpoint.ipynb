# Religion analysis

This project focuses on the analysis of global religious patterns using multiple international datasets. The analysis combines demographic, geographic, and socioeconomic perspectives in order to better understand how religion differs across countries and how it relates to various social indicators.

The project uses four main datasets. The dataset number-of-people-by-religion.csv contains estimates of the number of people belonging to major religions in different countries. The dataset how-important-religion-is-in-your-life.csv describes how important religion is in people’s lives based on survey responses. Additional socioeconomic information such as GDP, fertility rate, income group, and population is taken from World_bank.csv, while life_expectancy_years.csv provides historical life expectancy data.

Several questions are explored throughout the analysis. The project examines which religions dominate different regions, how the global religious structure changed between 2010 and 2020, and how important religion is in different societies. In addition, the analysis investigates possible relationships between religiousness and socioeconomic indicators such as GDP, fertility rate, life expectancy, and income level.

We have a problem in that the columns containing the number of people of a given faith may contain both very small values and “< 10,000” 
The question is what to do with them: 
1) Replace them with something around 10,000 so we can perform calculations with them later, but then there’s a risk that we’ll distort the information
2) Keep the “<10,000” category, but then we can’t perform any calculations with these values
3) Replace all values with categories, sort of smoothing them out, but we can’t do that either, since populations in different countries may vary, and 10,000 people in a country with a population of one million and 40 million need to be coded into different categories

There’s no single right answer here, but we decided to set the values to 10,000 because the data won’t be distorted too much, and the calculations will clearly be very important to us

```python
number_of_people = number_of_people.replace('< 10000', 10000)

religions = ['Buddhists', 'Hindus', 'Muslims', 'Jews', 'Christians', 'No religion', 'Other religions']

number_of_people[religions] = (
    number_of_people[religions]
    .replace('< 10000', 10000)
    .astype(int)
)
```

### 2.2 Global distribution of religions

#### How has the number of adherents of specific religions changed over the past 10 years (from 2010 to 2020)?

To compare the global distribution of religions between 2010 and 2020, the share of each religious group in the world population is calculated. The percentage share is computed as:

$$
p_i = \frac{x_i}{\sum_{j=1}^{n} x_j} \times 100
$$

where:

- $x_i$ represents the number of people belonging to the $i$-th religion,
- $n$ is the total number of religious groups.

This transformation makes it possible to compare religions proportionally instead of using absolute population counts, which can differ significantly between years.

```python
world_2010 = number_of_people_world[
    number_of_people_world['Year'] == 2010
]

world_2020 = number_of_people_world[
    number_of_people_world['Year'] == 2020
]

comparison = pd.DataFrame({
    'Religion': religions,
    '2010': world_2010[religions].iloc[0].values,
    '2020': world_2020[religions].iloc[0].values
})

comparison = comparison.melt(
    id_vars='Religion',
    var_name='Year',
    value_name='People'
)

comparison['Percent'] = comparison.groupby('Year')['People'].transform(
    lambda x: x / x.sum() * 100
)

comparison['Year'] = comparison['Year'].astype(str)

order = (
    comparison[comparison['Year'] == '2020']
    .sort_values('Percent', ascending=False)['Religion']
)

comparison.head()
```

To analyze changes in the global religious structure, the percentage shares of religions in 2010 and 2020 are compared. The change is calculated as the difference between the two years:

$$
\Delta p_i = p_{i,2020} - p_{i,2010}
$$

```python
changes = (
    comparison
    .groupby(['Religion', 'Year'])['Percent']
    .sum()
    .unstack()
)

changes['Change'] = changes['2020'] - changes['2010']

changes = changes.sort_values('2020', ascending=False)

changes
```

```python
fig, ax = plt.subplots(figsize=(12, 6))

sns.barplot(
    data=comparison,
    x='Religion',
    y='Percent',
    hue='Year',
    order=order,
    ax=ax
)

for container in ax.containers:
    ax.bar_label(
        container,
        fmt='%.1f%%',
        padding=3,
        fontsize=12,
        fontweight='bold'
    )

ax.set_xlabel('')
ax.set_ylabel('')
ax.set_yticks([])

plt.xticks(
    rotation=20,
    fontsize=11,
    fontweight='bold'
)

plt.title(
    'Distribution of Religions in the World',
    fontsize=18,
    fontweight='bold'
)

plt.legend(
    title='Year',
    fontsize=12,
    title_fontsize=13
)

save_path = FIG / 'religions_world_barplot.png'

plt.savefig(
    save_path,
    dpi=300,
    bbox_inches='tight'
)

print(f'Figure saved to: {save_path}')

plt.show()
```

### 2.3 Religious diversity

#### 2.3.1 Diversity index

Religious diversity can be understood as the distribution of a population across multiple religious groups. To measure the heterogeneity of such a distribution, entropy is used because it takes into account both the number of religions and the evenness of their distribution.

Entropy is defined as:

$$
H = - \sum_{i=1}^{n} p_i \log(p_i)
$$

where:

- $p_i$ represents the share of the $i$-th religion in a country,
- $n$ is the number of religious groups.

A low entropy value indicates that one religious group strongly dominates the population. In contrast, a high entropy value means that the population is distributed more evenly across multiple religions.

If one religious group represents $100\%$ of the population, then:

$$
H = 0
$$

which corresponds to a completely homogeneous society. The maximum entropy value occurs when all religions have equal shares within the population.

```python
religion_diversity = number_of_people[
    number_of_people["Code"].notna() &
    (number_of_people["Code"].str.len() == 3) &
    (~number_of_people["Code"].str.startswith("OWID"))
].copy()

religion_diversity["population"] = religion_diversity[religions].sum(axis=1)

shares = religion_diversity[religions].div(
    religion_diversity["population"],
    axis=0
)

shares_log = shares.where(shares > 0)

religion_diversity["entropy"] = -(
    shares * np.log(shares_log)
).sum(axis=1)

religion_diversity = religion_diversity[
    ["Entity", "Code", "Year", "population", "entropy"] + religions
]

religion_diversity.head()
```

#### 2.3.2 Diversity maps

```python
fig, axes = plt.subplots(2, 1, figsize=(24, 16))

years = [2010, 2020]

vmin = religion_diversity['entropy'].min()
vmax = religion_diversity['entropy'].max()

for ax, year in zip(axes, years):

    data_year = religion_diversity[
        religion_diversity['Year'] == year
    ].copy()

    merged = world.merge(
        data_year[['Code', 'entropy']],
        left_on='ISO_A3',
        right_on='Code',
        how='left'
    )

    merged.plot(
        column='entropy',
        cmap='YlOrRd',
        linewidth=0.5,
        edgecolor='black',
        legend=True,
        ax=ax,
        vmin=vmin,
        vmax=vmax,
        missing_kwds={'color': 'lightgrey'},
        legend_kwds={'shrink': 0.55}
    )

    ax.set_xlim(-180, 190)
    ax.set_ylim(-58, 88)

    ax.set_title(
        f'Religious Diversity Entropy, {year}',
        fontsize=22,
        fontweight='bold',
        pad=20
    )

    ax.axis('off')

plt.subplots_adjust(
    left=0.01,
    right=0.93,
    top=0.94,
    bottom=0.02,
    hspace=0.08
)

save_path = FIG / 'religious_diversity_entropy_maps_2010_2020.png'

plt.savefig(
    save_path,
    dpi=300,
    bbox_inches='tight'
)

print(f'Figure saved to: {save_path}')

plt.show()
```

#### 2.3.4 Diversity vs dominance

To compare religious diversity with the dominance of the largest religious group, the share of the dominant religion is calculated for each country.

The dominant religion share is defined as:

$$
D = \frac{\max(x_1, x_2, \dots, x_n)}{\sum_{i=1}^{n} x_i} \times 100
$$

where:

- $x_i$ represents the number of people belonging to the $i$-th religion,
- $n$ is the number of religious groups.

Higher values indicate that one religion strongly dominates the population, while lower values suggest a more balanced religious structure.

```python
dominance = religion_diversity.copy()

dominance['main_religion'] = dominance[religions].idxmax(axis=1)

dominance['main_religion_people'] = dominance[religions].max(axis=1)

dominance['main_religion_share'] = (
    dominance['main_religion_people'] / dominance['population'] * 100
)

dominance = dominance[
    ['Entity', 'Code', 'Year', 'population', 'entropy',
     'main_religion', 'main_religion_share']
]

dominance.head()
```

```python
dominance.sort_values('entropy')
```

```python
years = [2010, 2020]

selected_countries = [
    'Afghanistan',
    'Saudi Arabia',
    'Pakistan',
    'Brazil',
    'India',
    'Russia',
    'China',
    'Slovakia',
    'Ukraine',
    'Czechia',
    'Spain',
    'Ghana',
    'United States',
    'Malaysia',
    'Nigeria',
    'Japan',
    'Canada',
    'France',
    'United Kingdom',
    'South Korea',
    'Singapore',
    'Germany'
]

label_offsets = {
    'Slovakia': (4, 4),
    'Czechia': (4, -5),
    'United Kingdom': (8, 8),
    'South Korea': (8, -2),
}

for year in years:

    fig, ax = plt.subplots(figsize=(15, 9))

    data_year = dominance[
        dominance['Year'] == year
    ].copy()

    large_countries = data_year[
        data_year['population'] > 30_000_000
    ].copy()

    small_countries = data_year[
        data_year['population'] <= 30_000_000
    ].copy()

    ax.scatter(
        small_countries['entropy'],
        small_countries['main_religion_share'],
        s=18,
        alpha=0.35,
        edgecolor='none'
    )

    ax.scatter(
        large_countries['entropy'],
        large_countries['main_religion_share'],
        s=large_countries['population'] / 4_000_000,
        alpha=0.65,
        edgecolor='black',
        linewidth=0.6
    )

    labels = data_year[
        data_year['Entity'].isin(selected_countries)
    ]

    for _, row in labels.iterrows():

        x_offset, y_offset = label_offsets.get(
            row['Entity'],
            (6, 4)
        )

        ax.annotate(
            row['Entity'],
            xy=(row['entropy'], row['main_religion_share']),
            xytext=(x_offset, y_offset),
            textcoords='offset points',
            fontsize=10,
            fontweight='bold'
        )

    ax.set_title(
        f'Religious Diversity vs Dominance ({year})',
        fontsize=22,
        fontweight='bold',
        pad=20
    )

    ax.set_xlabel(
        'Religious diversity entropy',
        fontsize=14,
        fontweight='bold'
    )

    ax.set_ylabel(
        'Dominant religion share (%)',
        fontsize=14,
        fontweight='bold'
    )

    ax.grid(alpha=0.25)

    ax.text(
    0.02,
    0.03,
    'Orange circles represent countries\nwith population above 30 million',
    transform=ax.transAxes,
    fontsize=11,
    bbox=dict(
        facecolor='white',
        alpha=0.8,
        edgecolor='gray'
    )
)

    save_path = FIG / f'religious_diversity_vs_dominance_{year}.png'

    plt.savefig(
        save_path,
        dpi=300,
        bbox_inches='tight'
    )

    print(f'Figure saved to: {save_path}')

    plt.show()
```

The figure shows a strong negative relationship between religious diversity and the share of the dominant religion. Countries with low entropy values are typically characterized by a single religion that strongly dominates the population, while countries with higher entropy values tend to have a more balanced religious structure.

Most countries are concentrated in the upper-left part of the figure, indicating that religious homogeneity is more common globally than highly diverse religious compositions. In contrast, countries with very high diversity represent relatively rare cases.

The distribution of points also suggests that many countries combine the presence of a dominant religion with substantial religious minorities. Extremely high diversity is generally associated with lower dominance of the largest religious group.
